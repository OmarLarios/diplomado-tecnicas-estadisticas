## Variable aleatoria uniforme discreta

Una variable aleatoria $X$ tiene distribución uniforme discreta en el conjunto $\{x_{1},\dots,x_{n}\}$ si su función de masa de probabilidades está dada por
$$p_{X}(k)=\mathbb{P}(X=k)=\begin{cases}
\frac{1}{n} &\quad \text{si } k\in\{x_{1},\dots,x_{n}\} \\
0 &\quad \text{en otro caso}
\end{cases}$$

**Notación:** $X\sim Unif(x_{1},\dots,x_{n}).$

Se llama uniforme porque cada uno de los posibles valores de $X$ tienen la misma probabilidad.

Si $X\sim Unif(x_{1},\dots,x_{n})$, en este caso, se tiene:
	$$\mathbb{E}[X] = \sum_{k}kp_{X}(k)=\frac{x_n + x_1}{2}.$$

```python
#Probabilidad de que en un día ocurran más de 2 accidentes
mu = 2
prob = 1 - poisson.cdf(k=2, mu=mu)
print("La probabilidad de que en 1 día ocurran más de 2 accidentes es: ", prob)
```

```python
#Probabilidad de que en un día ocurran más de 2 accidentes
#sabiendo que ocurre por lo menos uno
prob1accidente = 1 - poisson.cdf(k=0, mu=mu)
prob2 = prob/prob1accidente
print("La probabilidad de que en 1 día ocurran más de 2 accidentes sabiendo que ocurre por lo menos 1 es: ", round(prob2,4))
```

```python
#Graficar la función de distribución
x = np.arange(0,9)
probabilidades = poisson.pmf(x, mu)
plt.figure(figsize=(8,5))
plt.bar(x, probabilidades, color="skyblue", edgecolor= "k")
plt.xlabel("accidentes por día")
plt.ylabel("probabilidad")
plt.title("Función de densidad de una Poisson")
plt.xticks(x)
plt.show()
```
